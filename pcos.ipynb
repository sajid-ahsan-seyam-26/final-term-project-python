# Data cleaning and preparation
1. column names are cleaned beacause original names have spaces and symbols
2. duplicate rows are removed 
3. numeric looking text and columns are converted inot numeric formet 
4. missing numeric values are filled with median because it can occurs extreme values exits
5. identifier columns are removed becaue serial numbers and patiendts file  numbers do not explain pcos patterns

```python
data=df.copy()
data.columns = (
    data.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)
#remove korbe duplicate rows gulake

before_rows = data.shape[0]
data = data.drop_duplicates()
after_rows = data.shape[0]

# convert korbe text columns to numbers

text_numeric_cols = ["ii_beta_hcg_miu_ml", "amh_ng_ml"]
for col in text_numeric_cols:
    data[col] = pd.to_numeric(data[col], errors="coerce")

    
# missing numeric values k median diye fill korbe

numeric_cols = data.select_dtypes(include=["int64", "float64"]).columns
for col in numeric_cols:
    if data[col].isnull().sum() > 0:
        data[col] = data[col].fillna(data[col].median())

# identifier columns drop korbe
data = data.drop(columns=["sl_no", "patient_file_no"], errors="ignore")

print("Duplicate rows removed:", before_rows - after_rows)
print("Cleaned dataset shape:", data.shape)
print("\nMissing values after cleaning:")
display(data.isnull().sum())
```

# Feature Engineering

```python
#  readable PCOS status
data["pcos_status"] = np.where(data["pcos_y_n"] == 1, "PCOS", "No PCOS")

#  total follicle count korbe
data["total_follicles"] = data["follicle_no_l"] + data["follicle_no_r"]

#  LH/FSH ratio ber korbe using NumPy
data["lh_fsh_ratio"] = np.where(
    data["fsh_miu_ml"] > 0,
    data["lh_miu_ml"] / data["fsh_miu_ml"],
    np.nan
)

#  age group korbe
data["age_group"] = pd.cut(
    data["age_yrs"],
    bins=[0, 25, 30, 35, 40, 100],
    labels=["<=25", "26-30", "31-35", "36-40", "40+"]
)

#  BMI group korbe
data["bmi_group"] = pd.cut(
    data["bmi"],
    bins=[0, 18.5, 25, 30, 100],
    labels=["Underweight", "Normal", "Overweight", "Obese"]
)

# symptom and lifestyle score
symptom_lifestyle_cols = [
    "weight_gain_y_n",
    "hair_growth_y_n",
    "skin_darkening_y_n",
    "hair_loss_y_n",
    "pimples_y_n",
    "fast_food_y_n"
]

data["symptom_lifestyle_score"] = data[symptom_lifestyle_cols].sum(axis=1)

# follicle level
data["follicle_level"] = pd.cut(
    data["total_follicles"],
    bins=[-1, 8, 16, 100],
    labels=["Low", "Medium", "High"]
)

display(data[[
    "pcos_status", "total_follicles", "lh_fsh_ratio",
    "age_group", "bmi_group", "symptom_lifestyle_score",
    "follicle_level"
]].head(100))
```

# Data Analysis

```python
# Analysis 1: PCOS count and percentage
pcos_count = data["pcos_status"].value_counts()
pcos_rate = data["pcos_y_n"].mean() * 100

print("PCOS count:")
display(pcos_count)

print("PCOS percentage:", round(pcos_rate, 2), "%")
```

```python
# Analysis 2: subgroup comparison by PCOS status
compare_by_pcos = data.groupby("pcos_status")[
    ["age_yrs", "bmi", "waist_hip_ratio", "total_follicles",
     "amh_ng_ml", "lh_fsh_ratio", "symptom_lifestyle_score"]
].mean().round(2)

display(compare_by_pcos)
```

```python
# Analysis 3: subgroup comparison by BMI group
pcos_by_bmi = (data.groupby("bmi_group", observed=False)["pcos_y_n"].mean() * 100).round(2)
display(pcos_by_bmi)
```

```python
# Analysis 4: subgroup comparison by age group
pcos_by_age = (data.groupby("age_group", observed=False)["pcos_y_n"].mean() * 100).round(2)
display(pcos_by_age)
```

```python
# Analysis 5: lifestyle and symptom comparison
binary_cols = [
    "fast_food_y_n",
    "reg_exercise_y_n",
    "weight_gain_y_n",
    "hair_growth_y_n",
    "skin_darkening_y_n",
    "hair_loss_y_n",
    "pimples_y_n"
]

for col in binary_cols:
    print("\nPCOS percentage by", col)
    result = (data.groupby(col)["pcos_y_n"].mean() * 100).round(2)
    display(result)
```

```python
# Analysis 6: PCOS percentage by follicle level
pcos_by_follicle_level = (data.groupby("follicle_level", observed=False)["pcos_y_n"].mean() * 100).round(2)
display(pcos_by_follicle_level)
```

```python
# Analysis 7: relationship analysis using correlation
corr_cols = [
    "pcos_y_n", "age_yrs", "bmi", "waist_hip_ratio", "total_follicles",
    "amh_ng_ml", "lh_fsh_ratio", "symptom_lifestyle_score",
    "bp_systolic_mmhg", "bp_diastolic_mmhg", "rbs_mg_dl"
]

correlation_with_pcos = (
    data[corr_cols]
    .corr(numeric_only=True)["pcos_y_n"]
    .drop("pcos_y_n")
    .sort_values(ascending=False)
    .round(3)
)

display(correlation_with_pcos)
```

```python
# Analysis 8: outlier analysis using NumPy z-score and percentiles
outlier_cols = [
    "fsh_miu_ml", "lh_miu_ml", "vit_d3_ng_ml",
    "rbs_mg_dl", "amh_ng_ml", "total_follicles", "bmi"
]

outlier_table = []

for col in outlier_cols:
    values = data[col].dropna().to_numpy()
    mean_value = np.mean(values)
    std_value = np.std(values)
    z_scores = (values - mean_value) / std_value

    outlier_count = np.sum(np.abs(z_scores) > 3)
    median_value = np.percentile(values, 50)
    p95_value = np.percentile(values, 95)
    max_value = np.max(values)

    outlier_table.append([
        col,
        int(outlier_count),
        round(median_value, 2),
        round(p95_value, 2),
        round(max_value, 2)
    ])

outlier_table = pd.DataFrame(
    outlier_table,
    columns=["column", "z_score_outliers", "median", "95th_percentile", "maximum"]
)

display(outlier_table)
```
